## Model training

```python
# dicvide the daat into x and y
X=df["Temperature (°C)"].values.reshape(-1,1)
y=df["Ice Cream Sales (units)"]
```

```python
# ploynomial regression

from sklearn.preprocessing import PolynomialFeatures

# create the insatance

x_poly=PolynomialFeatures(degree=2)

# fit and transfrom 
x_poly=x_poly.fit_transform(X)
```

```python
x_poly
```

```python
# model building
from sklearn.linear_model import LinearRegression

# cretae th einstance of the model
LR=LinearRegression()

# train the model

model=LR.fit(x_poly,y)

# model rediction

y_pred=model.predict(x_poly)

```

```python
# model evaluation metrics

from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error,mean_squared_error,r2_score

# MSE
MSE=mean_squared_error(y,y_pred)

# MAE
MAE=mean_absolute_error(y,y_pred)

# r2
r2=r2_score(y,y_pred)

# print
print("MSE:      ",MSE)
print("MAE:      ",MAE)
print("R2_score: ",r2)
```
